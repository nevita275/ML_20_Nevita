# Persiapan & Load Data

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('wbc.csv')
df.head()
```

# Tugas No.1 - Seleksi Fitur

```python
# Membuang kolom 'id' karena tidak berguna untuk prediksi
# Kita gunakan errors='ignore' agar tidak error jika kolom tersebut tidak ada
df = df.drop(columns=['id', 'Unnamed: 32'], errors='ignore')

print("Data setelah seleksi fitur:")
df.head()
```

# Tugas No.2 - Encoding Kolom "diagnosis"

```python
le = LabelEncoder()

# Proses encoding mengubah teks 'M' dan 'B' menjadi angka
df['diagnosis'] = le.fit_transform(df['diagnosis'])

print("\nData setelah Encoding (Perhatikan kolom diagnosis):")
df.head()
```

# Tugas No.3 - Standarisasi Nilai Numerik

```python
std = StandardScaler()

# Mengambil semua nama kolom numerik selain 'diagnosis'
numeric_cols = df.columns.drop('diagnosis')

# Melakukan standarisasi hanya pada kolom-kolom numerik tersebut
df[numeric_cols] = std.fit_transform(df[numeric_cols])

print("\nData akhir setelah Standarisasi:")
df.head()
```
